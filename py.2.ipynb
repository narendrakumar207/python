```python
#list in python
#list is a ordered and changeable collection that can store multiple values
marks = [80,50,75,85]

print(marks[0]) = 80
print(marks[1]) = 50
print(marks[2]) = 75
print(marks[3]) = 85
```

```python
#change elements in a list 
marks = [80,50,75]

marks[1] = 95 
marks[0] 
marks[2]
marks[3]
print(marks)
```

```python
#change elements in a list 
marks = [80,50,75]
marks[1] = 95
print(marks)
```

```python
#add elements to a list 
marks = [80,90,75]

marks.append (85)
print(marks)
```

```python
#remove elements from a list
marks = [80,90,75]
marks.remove(90)
print(marks)
```

```python
#extend a list 
a=[1,2,3]
b=[4,5,6]
a.extend(b)
print(a)    
```
