```python
#for loop 

#used to repeat code or iterate a sequence.
for i in range(1,6):
    print(i)
```

```python
#use while when repetition depends on a condition.
#looping through number 1to 5 using while loop 
i = 1

while i<=5:
    print(i)
    i = i + 1
```

```python
 #sum of number from 1 to n
n= int(input ("enter n:"))

total =0
for i in range(1, n +1):
    total  = total + i
    print ("sum:",total)
```

```python
#factorial of a number
n = int(input("enter number:"))

factorial = 1

for i in range(1, n +1):
    factorial = factorial * i

print("factorial:", factorial)
```
